```python
from dotenv import load_dotenv
import requests
import os

loaded = load_dotenv()

if not loaded:
    raise RuntimeError(".env file not found")
```

```python
from datetime import datetime

FOLDER_NAME = "requests"


def make_requests(path: str, params: dict | None = None) -> requests.Response:
    response = requests.get(
        os.getenv("API_FOOTBALL_URL") + path,
        headers={
            "x-apisports-key": os.getenv("API_FOOTBALL_API_KEY"),
            "x-rapidapi-host": os.getenv("API_FOOTBALL_URL")
            .replace("https://", "")
            .replace("/", ""),
        },
        params=params if params else None,
    )
    response.raise_for_status()

    # log all calls to a request folder
    os.makedirs(FOLDER_NAME, exist_ok=True)
    now = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")

    with open(
        os.path.join(FOLDER_NAME, f"{path.replace('/', '_')}_{now}.txt"),
        "w",
        encoding="utf-8",
    ) as f:
        f.write(f"URL: {response.url}\n")
        f.write(f"Status code: {response.status_code}\n")
        f.write(f"Response: {response.text}\n")

    return response
```

```python
status_response = make_requests("/status", {})
print(status_response.json())
```
